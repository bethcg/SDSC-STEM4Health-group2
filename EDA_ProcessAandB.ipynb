## Data Visualization

```python
import pandas as pd
from pandas.api.types import is_numeric_dtype

import pandas as pd
import matplotlib.pyplot as plt

def plot_distributions(df, bins=30):
    for col in df.columns:
        plt.figure(figsize=(8, 4))

        df[col].hist(bins=bins, edgecolor="black")

        plt.title(f"Distribution of {col}")
        plt.xlabel(col)
        plt.ylabel("Number of batches")
        plt.tight_layout()
        plt.show()

```

```python
plot_distributions(df_process_B_calibration_data)
```

```python
plot_distributions(df_process_A_calibration_data)
```
