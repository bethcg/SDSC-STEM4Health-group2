# Process A: variable overview and PCA

Read-only analysis of the calibration data (the input files are not modified). Same approach as `pca_process_B.ipynb`.

* `calibration_data.csv`: 161 batches × 11 process variables. **Batches have different lengths** (about 200–2900 time steps).
* `calibration_labels.csv`: one label per batch (accepted / rejected)

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import roc_auc_score

DATA_DIR = "../input-data/Project 2 - BRPMM/process A/calibration data/"
df = pd.read_csv(DATA_DIR + "calibration_data.csv")
labels = pd.read_csv(DATA_DIR + "calibration_labels.csv").set_index("batch_id")["label"]

df["t"] = df.groupby("batch_id").cumcount()          # time step within each batch
VARS = [c for c in df.columns if c not in ("batch_id", "t")]
lengths = df.groupby("batch_id").size()
print(df.shape, "| batches:", df.batch_id.nunique())
print(labels.value_counts())
print("\nBatch length by label:"); print(lengths.groupby(labels.loc[lengths.index]).describe().round(0))
```

## 1. Each variable plotted separately
Each line is one batch over time (raw time steps). Grey = accepted, red = rejected.

```python
COLORS = {"accepted": "#9a9a9a", "rejected": "#d62728"}
step = 5

fig, axes = plt.subplots(3, 4, figsize=(20, 13))
for ax, var in zip(axes.flat, VARS):
    for bid, g in df.groupby("batch_id"):
        lab = labels[bid]
        ax.plot(g.t.values[::step], g[var].values[::step], color=COLORS[lab],
                lw=0.8 if lab == "rejected" else 0.5, alpha=0.6 if lab == "rejected" else 0.4,
                zorder=3 if lab == "rejected" else 2)
    ax.set_title(var); ax.set_xlabel("time step")
ax = axes.flat[len(VARS)]
ax.hist([lengths[labels.loc[lengths.index] == l] for l in COLORS], bins=30, color=list(COLORS.values()), label=list(COLORS), stacked=True)
ax.set_title("batch length"); ax.set_xlabel("number of time steps"); ax.legend()
fig.tight_layout(); plt.show()
```

A few short batches and brief start-up/shut-down spikes stretch the axes. Below, the same plots are clipped to the 1st–99th percentile range so the normal operating band is easier to see.

```python
fig, axes = plt.subplots(3, 4, figsize=(20, 13))
for ax, var in zip(axes.flat, VARS):
    for bid, g in df.groupby("batch_id"):
        lab = labels[bid]
        ax.plot(g.t.values[::step], g[var].values[::step], color=COLORS[lab],
                lw=0.8 if lab == "rejected" else 0.5, alpha=0.6 if lab == "rejected" else 0.4,
                zorder=3 if lab == "rejected" else 2)
    lo, hi = df[var].quantile([0.01, 0.99]); pad = (hi - lo) * 0.1
    ax.set_ylim(lo - pad, hi + pad); ax.set_title(var + " (zoomed)"); ax.set_xlabel("time step")
axes.flat[-1].axis("off")
fig.tight_layout(); plt.show()
```

## 2. PCA on the full trajectories (batch-wise unfolded / multiway PCA)

Because batch lengths differ, each batch is first resampled onto a common **normalised time axis** (0 = start, 1 = end) with 101 points, using linear interpolation. Each batch then becomes one row (11 variables × 101 points = 1111 columns). Each column is standardised, then PCA is fitted.

PCA is **unsupervised**, so for each component we check how well its score separates the labels (ROC AUC: 0.5 = none, 1.0 = perfect).

```python
N_PTS = 101
grid = np.linspace(0, 1, N_PTS)
batches = lengths.index
rows = []
for bid in batches:
    g = df[df.batch_id == bid]
    tn = np.linspace(0, 1, len(g))
    rows.append(np.concatenate([np.interp(grid, tn, g[v].values) for v in VARS]))
cols = pd.MultiIndex.from_product([VARS, range(N_PTS)])
X_unf = pd.DataFrame(np.array(rows), index=batches, columns=cols)
y = (labels.loc[batches] == "rejected").astype(int).values
Xs = StandardScaler().fit_transform(X_unf.values)

pca = PCA(n_components=10).fit(Xs)
scores = pca.transform(Xs)
auc = lambda s: max(roc_auc_score(y, s), 1 - roc_auc_score(y, s))
summary = pd.DataFrame({"explained_var": pca.explained_variance_ratio_.round(3),
                        "cum_explained": pca.explained_variance_ratio_.cumsum().round(3),
                        "AUC_vs_label": [round(auc(scores[:, k]), 3) for k in range(10)]},
                       index=[f"PC{k+1}" for k in range(10)])
summary
```

```python
best = summary.AUC_vs_label.values.argsort()[::-1][:2]
fig, axes = plt.subplots(1, 2, figsize=(14, 6))
lab_arr = labels.loc[batches].values
for ax, (i, j) in zip(axes, [(0, 1), tuple(sorted(best))]):
    for lab, c in COLORS.items():
        m = lab_arr == lab
        ax.scatter(scores[m, i], scores[m, j], c=c, label=lab, s=40, edgecolor="k", alpha=0.8)
    ax.set_xlabel(f"PC{i+1} ({pca.explained_variance_ratio_[i]:.0%})")
    ax.set_ylabel(f"PC{j+1} ({pca.explained_variance_ratio_[j]:.0%})")
    ax.legend()
axes[0].set_title("Top-2 variance components"); axes[1].set_title("Two components that best separate the labels")
plt.tight_layout(); plt.show()
```

Each variable's contribution to a component is the sum of its squared loadings over all its time points. Each component's contributions add up to 1.

```python
load = pd.DataFrame(pca.components_.T, index=X_unf.columns, columns=summary.index)
contrib = (load ** 2).groupby(level=0).sum().loc[VARS]
show = ["PC1", "PC2", "PC3"] + [summary.index[k] for k in best if summary.index[k] not in ("PC1", "PC2", "PC3")]
fig, ax = plt.subplots(figsize=(9, 6))
im = ax.imshow(contrib[show].values, cmap="Blues", aspect="auto")
ax.set_xticks(range(len(show)), [f"{p}\nAUC={summary.loc[p,'AUC_vs_label']}" for p in show])
ax.set_yticks(range(len(VARS)), VARS)
for r in range(len(VARS)):
    for c in range(len(show)):
        ax.text(c, r, f"{contrib[show].values[r, c]:.2f}", ha="center", va="center", fontsize=8)
fig.colorbar(im, label="share of component"); ax.set_title("Variable contribution to PCs")
plt.tight_layout(); plt.show()
```

## 3. PCA on per-batch summary features

Each variable is summarised per batch by its mean, standard deviation, min, max and final value. Batch length is added as an extra feature. PCA is then fitted on these features.

```python
summ = df.groupby("batch_id")[VARS].agg(["mean", "std", "min", "max", "last"])
summ.columns = [f"{v}__{s}" for v, s in summ.columns]
summ["batch_length"] = lengths
summ = summ.loc[batches]
Zs = StandardScaler().fit_transform(summ.values)
pca2 = PCA(n_components=10).fit(Zs); sc2 = pca2.transform(Zs)
summary2 = pd.DataFrame({"explained_var": pca2.explained_variance_ratio_.round(3),
                         "AUC_vs_label": [round(auc(sc2[:, k]), 3) for k in range(10)]},
                        index=[f"PC{k+1}" for k in range(10)])
display(summary2)
k_best = summary2.AUC_vs_label.values.argmax()
ld = pd.Series(pca2.components_[k_best], index=summ.columns)
print(f"Top loadings of PC{k_best+1} (best separating component):")
ld.reindex(ld.abs().sort_values(ascending=False).index).head(10).round(3)
```

## 4. Direct check: how well does each variable separate the labels on its own?

PCA ranks by **variance**, not by usefulness for the **label**. As a supervised sanity check: the ROC AUC of each single summary feature, plus each variable's best feature.

```python
uni = pd.Series({c: auc(summ[c].values) for c in summ.columns}).sort_values(ascending=False)
top = uni.head(15)
fig, axes = plt.subplots(1, 2, figsize=(16, 6))
axes[0].barh(top.index[::-1], top.values[::-1], color="#1f77b4"); axes[0].axvline(0.5, color="k", ls="--", lw=1)
axes[0].set_xlabel("ROC AUC (single feature vs. label)"); axes[0].set_xlim(0.4, 1.0); axes[0].set_title("Top 15 single features")
per_var = uni.groupby(lambda c: c.split("__")[0]).max().sort_values()
axes[1].barh(per_var.index, per_var.values, color="#1f77b4"); axes[1].axvline(0.5, color="k", ls="--", lw=1)
axes[1].set_xlim(0.4, 1.0); axes[1].set_xlabel("best single-feature AUC"); axes[1].set_title("Best feature per variable")
plt.tight_layout(); plt.show()
top.round(3)
```

```python
best_var = per_var.index[-1]
fig, axes = plt.subplots(1, 2, figsize=(15, 5))
for lab, c in COLORS.items():
    m = lab_arr == lab
    mean = X_unf[best_var].values[m].mean(0); sd = X_unf[best_var].values[m].std(0)
    axes[0].plot(grid, mean, color=c, label=f"{lab} mean"); axes[0].fill_between(grid, mean - sd, mean + sd, color=c, alpha=0.2)
axes[0].set_title(f"{best_var}: mean ± 1 sd by label (normalised time)"); axes[0].set_xlabel("normalised time"); axes[0].legend()
f = uni.index[0]
axes[1].boxplot([summ[f][lab_arr == l] for l in COLORS], tick_labels=list(COLORS)); axes[1].set_title(f)
plt.tight_layout(); plt.show()
```
